```kotlin
import dev.functionalml.ad.*
import kotlin.math.PI
```

## $f(x) = x^2 + 3x$

```kotlin
val f: (Dual) -> Dual = { x -> x * x + 3.0 * x }
val x = 2.0
val result = valueAndGrad(f, x)

mapOf("x" to x, "f(x)" to result.value, "f'(x)" to result.derivative)
```

```kotlin
val df = grad(f)
listOf(0.0, 1.0, 2.0, 3.0).associateWith(df)
```

## 기본 함수와 연쇄 법칙

```kotlin
val sinGrad = grad { value -> sin(value * value) }
val expGrad = grad { value -> exp(2.0 * value) }

mapOf("d/dx sin(x^2), x=pi" to sinGrad(PI), "d/dx exp(2x), x=1" to expGrad(1.0))
```
